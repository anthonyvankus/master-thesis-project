```python
import numpy as np
import mbtr.utils as ut
from scipy.stats import expon
import matplotlib.pyplot as plt
from mbtr.mbtr import MBT
from mbtr.utils import set_figure
from mbtr import LOSS_MAP
from CopulaLoss import CopulaLikelihood

LOSS_MAP['copula'] = CopulaLikelihood # this line has to be added for the MBT class to recognize the CopulaLikelihood subclass as a type of loss
```

```python
loss = CopulaLikelihood()
y_test = np.array([[0.1, 2.5],
                [1.3, 0.8]])
```

```python
#Test: set_dimension()
loss.set_dimension(10)
assert loss.n_dims == 3, "Dimension should be 3"
```

```python
#Test: get_intial_guess()
y_guess = loss.get_initial_guess(y_test)
correct_guess = np.array([[10/7, 20/33, 0.],
                          [10/7, 20/33, 0.]])
print(y_guess)
print(correct_guess)
#assert (y_guess==correct_guess).all(), "Initial guess is computed incorrectly"
```

```python
#Test: exp_lpdf_grad()
y_test0 = y_test[:, 0].reshape(-1,1)
par_test0 = np.array([[1], [2]])
exp_grad_test = loss.exp_lpdf_grad(y_test0, par_test0)
print(exp_grad_test) #0.9, -0.8

#Test: exp_lpdf_hess()
exp_hess_test = loss.exp_lpdf_hess(par_test0)
print(exp_hess_test) #-1.0, -0.25
```

```python
#Test: fgm_copula_lpdf()
u_test = np.array([[0.1, 0.5],
                [0.3, 0.8]])
par_test1 = np.array([[0.25], [0.8]])
fgm_lpdf_test = loss.fgm_copula_lpdf(u_test, par_test1)
print(fgm_lpdf_test) #0.0 + ln(0.808) = -0.21319...

#Test: fgm_copula_lpdf_grad()
fgm_grad_test = loss.fgm_copula_lpdf_grad(u_test, par_test1)
print(fgm_grad_test) #0.0, -0.24/0.808 = -0.297029...

#Test: fgm_copula_lpdf_hess()
fgm_hess_test = loss.fgm_copula_lpdf_hess(u_test, par_test1)
print(fgm_hess_test) #0.0, -0.088226644...

#Test: fgm_par_link()
fgm_link_test = loss.fgm_par_link(par_test1)
print(fgm_link_test) #0.41421..., 3.0776...

#Test: fgm_par_inv_link()
fgm_inv_link_test = loss.fgm_par_inv_link(fgm_link_test)
print(fgm_inv_link_test) #0.25, 0.8

#Test: fgm_par_inv_link_grad()
fgm_inv_link_grad_test = loss.fgm_par_inv_link_grad(fgm_link_test)
print(fgm_inv_link_grad_test) #0.54338..., 0.06079...

#Test: fgm_par_inv_link_hess()
fgm_inv_link_hess_test = loss.fgm_par_inv_link_hess(fgm_link_test)
print(fgm_inv_link_hess_test) #-0.38423..., -0.035735...
```
